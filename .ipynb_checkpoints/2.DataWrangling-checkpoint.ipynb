## Step 2: Data Wrangling

This notebook takes the files we built in Step 1 and turns them into one clean table we can use to train models1. 
- Read **`Nhanes_Merged_Raw.csv`** for the numbers instead of the `Merged_Decoded` sheet.The CSV keeps the raw codes.
- Load the **decoded** sheet and mirror every cleaning step onto it, so the final output has both a decoded tab and a raw tab.
- Drop **cycle-specific columns**
- Depression_Flag` is cast to a clean integer `0`/`1`, and declares which columns must not be used as features.

#### Import Libraries

```python
import os
import pandas as pd
import numpy as np
```

```python
#pip install xlsxwriter
```

#### Data Path

```python
BASE_DIR = "."
RAW_DIR      = os.path.join(BASE_DIR, "Nhanes_RawData")
MERGED_CSV   = os.path.join(RAW_DIR, "Nhanes_Merged_Raw.csv")   # NEW: raw numeric codes from Step 1
MAPPED_XLSX  = os.path.join(RAW_DIR, "Nhanes_Mapped.xlsx")      # codebook + decoded labels
CLEAN_CSV    = os.path.join(RAW_DIR, "Nhanes_Cleaned.csv")      # model-ready, raw codes
CLEAN_XLSX   = os.path.join(RAW_DIR, "Nhanes_Cleaned.xlsx")     # NEW: 2 tabs (decoded + raw)
```

#### Load the dataWe load three things. All the *calculations* are done on `raw` (numeric codes); `decoded` is carriedalong purely so the human-readable labels survive into the final output.

```python
codebook = pd.read_excel(MAPPED_XLSX, sheet_name="Codebook")
raw      = pd.read_csv(MERGED_CSV, low_memory=False)              # numeric codes - we compute on this
decoded  = pd.read_excel(MAPPED_XLSX, sheet_name="Merged_Decoded")  # text labels  - we mirror onto this

print("codebook rows:", codebook.shape)
print("raw data shape:", raw.shape)
print("decoded shape :", decoded.shape)

# the two frames must line up row-for-row or mirroring the cleaning would scramble the labels
decoded = decoded.set_index("SEQN").reindex(raw["SEQN"]).reset_index()
assert (decoded["SEQN"].values == raw["SEQN"].values).all(), "SEQN order mismatch"
print("raw and decoded are aligned row-for-row.")

raw.head()
```

#### Step 2.1: Clean up 

NHANES uses numbers like `7`, `9`, `77`, `99` to mean "Refused" or "Don't know" instead of a real answer. If we leave these in, a model might think someone who "refused" scored super high on a question. So the first real cleaning step is: for every column, look up (in the codebook) which codes mean Refused/Don't know, and turn those into `NaN` (missing) instead.
This step records the 4- and 5-digit sentinel codes used on the *continuous* questions (`PAD680 = 9999`, `SMQ050Q = 99999`, etc.), so this step catches many more bad values than before - without it a model would read "Don't know" as 9,999 minutes of sitting. We also mirror each blanking onto the decoded frame so both versions stay in sync. Note we deliberately keep codes like `SLD012 = 2` ("Less than 3 hours") - those are real answers stored as a floor value, not missing data.

```python
# build a dictionary: {column_name: set of codes that mean "Refused" or "Don't know"}
missing_code_map = {}
for _, row in codebook.iterrows():
    if row["Meaning"] in ("Refused", "Don't know"):
        missing_code_map.setdefault(row["Variable"], set()).add(row["Code"])

print(f"Found Refused/Don't know codes for {len(missing_code_map)} columns.")

df = raw.copy()
before = df.isna().sum().sum()
for col, codes in missing_code_map.items():
    if col in df.columns:
        df[col] = df[col].replace(list(codes), np.nan)
print(f"Blanked {df.isna().sum().sum() - before:,} sentinel values.")

# mirror onto the decoded frame: wherever raw just became NaN, blank the label too
dec = decoded.copy()
for col in missing_code_map:
    if col in dec.columns and col in df.columns:
        dec[col] = dec[col].where(df[col].notna())

print("Done. Data shape is unchanged (still just replacing values):", df.shape)
```

#### Step 2.2: Build our target: a depression flag from the PHQ-9

The PHQ-9 is 9 questions (`DPQ010`–`DPQ090`), each scored 0-3 (higher = worse). Adding all 9 up gives a score from 0 to 27. This is a well-known, standard screening tool, and a score of **10 or more is the standard cutoff** used in research to flag likely depression. Rule we use: if someone is missing *any* of the 9 answers, we can't trust their total score, so their score (and flag) is left as missing rather than guessed.

```python
phq_cols = [f"DPQ0{n}0" for n in range(1, 10)]  
print("PHQ-9 columns:", phq_cols)

answered_all_9 = df[phq_cols].notna().all(axis=1)
df["PHQ9_Score"] = df[phq_cols].sum(axis=1, min_count=9)   # min_count=9 -> NaN unless all 9 present
df.loc[~answered_all_9, "PHQ9_Score"] = np.nan

# Standard clinical cutoff: score >= 10 means likely depression
df["Depression_Flag"] = np.where(
    df["PHQ9_Score"].isna(), np.nan,
    np.where(df["PHQ9_Score"] >= 10, 1, 0)
)

print(df["PHQ9_Score"].describe())
print()
print("Depression_Flag counts (1 = likely depression, 0 = not, NaN = we can't tell):")
print(df["Depression_Flag"].value_counts(dropna=False))
```

#### Step 2.3: Keep only rows where we know the answer

We can only train a model on people where we actually know if they show depression indicators or not, so we drop anyone with a missing `Depression_Flag`.**Amended:** the flag is now cast to a clean integer `0`/`1` (it was coming out as `0.0`/`1.0` floats because of the `NaN`s), and the decoded frame is filtered to the same rows.

```python
keep_rows = df["Depression_Flag"].notna()

clean = df.loc[keep_rows].copy()
cdec  = dec.loc[keep_rows].copy()                       # keep the decoded twin in step

clean["Depression_Flag"] = clean["Depression_Flag"].astype(int)   # 0/1, not 0.0/1.0
cdec["PHQ9_Score"]       = clean["PHQ9_Score"].values
cdec["Depression_Flag"]  = clean["Depression_Flag"].values

print("Rows before:", df.shape[0], " -> Rows after keeping only a known target:", clean.shape[0])
print()
print("Rows by cycle:")
print(clean["CYCLE"].value_counts())
print()
print("Class balance (as %):")
print((clean["Depression_Flag"].value_counts(normalize=True) * 100).round(1))
print()
print("Prevalence by cycle (%):")
print((clean.groupby("CYCLE")["Depression_Flag"].mean() * 100).round(2))
```

**Note:** only about 1 in 9 people are flagged as likely depression. This is an *imbalanced* target, keep this in mind for the modeling notebook (e.g. use `stratify=` when splitting train/test, and don't rely on plain accuracy alone, precision/recall/F1 matter more here).**Also note** the gap between the two cycles: roughly 9% pre-pandemic vs 13% in 2021-2023. That is a real period effect, and it is why the extra drop rule in Step 2.5 matters.

#### Step 2.4: Check how much data is missing in each columnSome columns (mostly ones with built-in skip logic, like smoking-detail questions asked only to smokers) are missing for most people.

A column that's missing for, say, 95% of people isn't very useful as a predictor, so we'll drop columns above a missingness threshold.
We now show missingness **per cycle** as well as overall. With two cycles pooled, a column can look "half missing" simply because it was only asked in one of them - the split view makes that obvious at a glance.

```python
missing_pct  = clean.isna().mean().sort_values(ascending=False) * 100
missing_cyc  = clean.groupby("CYCLE").apply(lambda g: g.isna().mean() * 100).T
missing_cyc.columns = [f"missing% {c}" for c in missing_cyc.columns]

missing_report = pd.concat([missing_pct.rename("percent_missing"), missing_cyc], axis=1)
missing_report.sort_values("percent_missing", ascending=False).round(1)
```

#### Step 2.5: Decide which columns to drop

We drop 5 kinds of columns:
- **Columns only asked in one cycle** (NEW - see the warning below).
- **Too much missing data** (more than 50% missing) — not reliable enough to use.
- **The PHQ-9 questions themselves** (`DPQ010`-`DPQ090`, `DPQ100`) — we used these to *build* our target, so using them again as predictors would be "cheating" (this is called data leakage — the model would just learn to re-read the target instead of finding real patterns).
   **Survey design/weighting columns** (`WTINT`, `WTMEC`, `WTINT2YR`, `WTMECPRP`, `SDMVSTRA`, `SDMVPSU`, `SDDSRVYR`, `RIDSTATR`) — these describe how NHANES samples people, not a person's health/lifestyle, so they aren't useful lifestyle predictors for this project. (`WTMEC_COMB` is kept but marked non-feature in Step 2.7, so weighted prevalence can still be reported.)
- **Redundant sleep clock-times** (`SLQ300`, `SLQ310`, `SLQ320`, `SLQ330`) — these are text times like "21:30", and we already have the same information as plain numbers of sleep hours in `SLD012`/`SLD013`. Same idea for `RIDAGEMN` (age in months) next to `RIDAGEYR`.>
-**Why the new "one cycle only" rule.** Once two cycles are pooled, the plain >50% rule stops measuring> *usefulness* and starts measuring *which cycle is bigger*. `SLQ050` (told a doctor about sleep trouble)> is `_P`-only and shows 39.8% missing, so the old rule **keeps** it. `DMQMILIZ` (military service) is> `_L`-only and shows 60.3% missing, so the old rule **drops** it. Every `_P` variable survives and every> `_L` variable dies, purely because the pre-pandemic cycle contributed more people.>> Worse: if we then fill `SLQ050` with `-1` for all 5,455 `_L` rows, that `-1` becomes a perfect flag for> "this row is from 2021-2023" — which is itself correlated with the target (13% vs 9%). The model would> score well by detecting the survey cycle rather than anything about the person. So we keep only the> **52 variables measured in both cycles**, using the `Available_In` column the codebook now carries.*(The old `PAD810Q` "orphaned unit" rule is no longer needed — `PAD810Q` is `_L`-only, so the cycle rule already removes it.)*

```python
# NEW: variables measured in both cycles (from the codebook's Available_In column)
shared_vars = set(codebook.loc[codebook["Available_In"] == "both", "Variable"])
BOOKKEEPING = ["SEQN", "CYCLE", "CYCLE_LABEL", "SOURCE_SUFFIX",
               "WTMEC_COMB", "PHQ9_Score", "Depression_Flag"]

cycle_specific_cols = [c for c in clean.columns
                       if c not in shared_vars and c not in BOOKKEEPING]

high_missing_cols = missing_pct[missing_pct > 50].index.tolist()
leakage_cols   = phq_cols + ["DPQ100"]
design_cols    = ["WTINT", "WTMEC", "WTINT2YR", "WTMEC2YR", "WTINTPRP", "WTMECPRP",
                  "WTINT_COMB", "SDMVSTRA", "SDMVPSU", "SDDSRVYR", "RIDSTATR",
                  "CYCLE_LABEL", "SOURCE_SUFFIX"]
redundant_cols = ["SLQ300", "SLQ310", "SLQ320", "SLQ330", "RIDAGEMN"]

drop_cols = (set(cycle_specific_cols) | set(high_missing_cols) | set(leakage_cols)
             | set(design_cols) | set(redundant_cols))
drop_cols = sorted(c for c in drop_cols if c in clean.columns)

def _why(c):
    if c in cycle_specific_cols: return "asked in one cycle only"
    if c in leakage_cols:        return "target leakage"
    if c in design_cols:         return "survey design / weight"
    if c in redundant_cols:      return "redundant"
    return "high missing"

print(f"Dropping {len(drop_cols)} columns:")
for c in drop_cols:
    print(f"   {c:12s} {_why(c)}")

model_df  = clean.drop(columns=drop_cols)
model_dec = cdec.drop(columns=[c for c in drop_cols if c in cdec.columns])
print()
print("Shape after dropping columns:", model_df.shape)
```

#### Step 2.6: Fill in the small amount of missing data that's left

Using the codebook, we already know which columns are **categorical** (a code stands for a category, like Male/Female) and which are **continuous** (a plain number, like age or hours of sleep). We fill missing values differently for each type:- **Continuous** columns → fill with the **median** (middle value), a common, simple choice that isn't thrown off by outliers.- **Categorical** columns → fill with `-1`, a brand new code meaning "not answered", so we don't accidentally pretend someone picked a real category they didn't.
The same fill is mirrored onto the decoded frame (categoricals get the text `"Not answered"` instead of `-1`). One caveat for the modeling notebook: the median here is taken over the whole dataset — if you want to be strict, re-fit the imputer on the training split only.

```python
continuous_vars = set(codebook.loc[codebook["Meaning"] == "continuous / numeric", "Variable"])
id_vars = set(codebook.loc[codebook["Meaning"] == "respondent ID", "Variable"])
categorical_vars = set(codebook["Variable"]) - continuous_vars - id_vars

remaining_continuous  = [c for c in model_df.columns if c in continuous_vars  and c not in BOOKKEEPING]
remaining_categorical = [c for c in model_df.columns if c in categorical_vars and c not in BOOKKEEPING]

print("Continuous columns to fill with median:", remaining_continuous)
print()
print("Categorical columns to fill with -1 ('not answered'):", remaining_categorical)

for col in remaining_continuous:
    med = model_df[col].median()
    model_df[col] = model_df[col].fillna(med)
    if col in model_dec.columns:
        model_dec[col] = model_dec[col].fillna(med)          # same number in both tabs

for col in remaining_categorical:
    model_df[col] = model_df[col].fillna(-1)
    if col in model_dec.columns:
        model_dec[col] = model_dec[col].fillna("Not answered")   # readable equivalent of -1

still_missing = model_df.drop(columns=[c for c in BOOKKEEPING if c in model_df]).isna().sum().sum()
print()
print(f"Missing values left in the whole table: {still_missing} (should be 0)")
```

#### Step 2.7: Quick sanity check on the final tableLet's take a final look before saving: shape, column list, and the first few rows.

We also declare `FEATURE_COLS` here and run a few automated checks.> **Important before you train.** `PHQ9_Score` is kept in the file as requested, but it must **not** be> fed to a model — `Depression_Flag` is defined as `PHQ9_Score >= 10`, so a model given the score scores> ~100% and has learned nothing. `CYCLE` is likewise kept for stratifying the train/test split and for> per-cycle reporting, not as an input. Use `X = model_df[FEATURE_COLS]` rather than "everything except> the target".

```python
NON_FEATURE = [
    "SEQN",             # respondent ID
    "CYCLE",            # stratify the split on this / report per cycle - NOT an input
    "WTMEC_COMB",       # survey weight - for weighted prevalence in the write-up - NOT an input
    "PHQ9_Score",       # TARGET LEAKAGE - the flag is derived from this
    "Depression_Flag",  # the target itself
]
TARGET       = "Depression_Flag"
FEATURE_COLS = [c for c in model_df.columns if c not in NON_FEATURE]

print("Final shape:", model_df.shape)
print()
print(f"{len(FEATURE_COLS)} features:", FEATURE_COLS)
print()
print("Non-feature columns kept in the file:", [c for c in NON_FEATURE if c in model_df.columns])
print()

def check(label, ok, detail=""):
    print(f"[{'PASS' if ok else 'FAIL'}] {label}" + (f"  -> {detail}" if detail else ""))

check("target is integer 0/1 only",
      set(model_df[TARGET].unique()) <= {0, 1} and model_df[TARGET].dtype.kind in "iu",
      f"{sorted(model_df[TARGET].unique())}, dtype={model_df[TARGET].dtype}")
check("no missing values among the features", model_df[FEATURE_COLS].isna().sum().sum() == 0)
check("no PHQ-9 item survived as a feature", not any(c.startswith("DPQ") for c in FEATURE_COLS))
check("PHQ9_Score is excluded from FEATURE_COLS", "PHQ9_Score" not in FEATURE_COLS)
check("every feature exists in both cycles", all(c in shared_vars for c in FEATURE_COLS))
check("both cycles represented", model_df["CYCLE"].nunique() == 2,
      model_df["CYCLE"].value_counts().to_dict())
check("decoded tab has the same rows",
      len(model_dec) == len(model_df) and (model_dec["SEQN"].values == model_df["SEQN"].values).all())

corr = (model_df[FEATURE_COLS].apply(pd.to_numeric, errors="coerce")
        .corrwith(model_df[TARGET]).abs().sort_values(ascending=False))
check("no single feature gives the target away", corr.max() < 0.9,
      f"strongest = {corr.index[0]} ({corr.iloc[0]:.2f})")

print()
print("Top features by correlation with the target:")
print(corr.head(8).round(3).to_string())
print()
w = model_df["WTMEC_COMB"].fillna(0)
print(f"Weighted prevalence (WTMEC_COMB): {np.average(model_df[TARGET], weights=w)*100:.2f}%"
      f"   |   unweighted: {model_df[TARGET].mean()*100:.2f}%")

model_df.head()
```

#### Step 2.8: Save the cleaned, model-ready table

This is the file the next notebook (model building) will load. `SEQN` is kept as a respondent ID (useful for joining more data later, or debugging) — it should be dropped or ignored as a *feature* before training, since it's just an ID number, not something that describes the person.**Amended:** we now save two things.- **`Nhanes_Cleaned.csv`** — raw numeric codes, the file the modeling notebook loads.- **`Nhanes_Cleaned.xlsx`** — two row-for-row identical tabs: `Cleaned_Decoded` (readable labels, for charts and the write-up) and `Cleaned_Raw` (the same numbers as the CSV, so the labels can always be re-derived later).

```python
model_df.to_csv(CLEAN_CSV, index=False)
print(f"Saved {CLEAN_CSV}")

with pd.ExcelWriter(CLEAN_XLSX, engine="xlsxwriter") as xl:
    model_dec.to_excel(xl, sheet_name="Cleaned_Decoded", index=False)   # readable labels
    model_df.to_excel(xl, sheet_name="Cleaned_Raw",      index=False)   # numeric codes
print(f"Saved {CLEAN_XLSX}  (tabs: Cleaned_Decoded, Cleaned_Raw)")

print()
print(f"Final cleaned dataset: {model_df.shape[0]} rows, {model_df.shape[1]} columns "
      f"({len(FEATURE_COLS)} usable features + {len(NON_FEATURE)} non-feature columns)")
```
