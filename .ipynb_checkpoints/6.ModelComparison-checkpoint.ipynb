```python
# Columns dropped by exact name - carried over unchanged from Step 5
DROP_EXACT = ["WTMEC_COMB"]

# Columns dropped by prefix, because one-hot encoding turned each one into many dummy columns
DROP_PREFIX = ["RIDRETH1_", "INDFMMPI", "INDFMMPC_", "ALQ142_"]

# Columns that must never reach a model: IDs, survey design, the score the target came from, the target itself
FORBIDDEN_COLUMNS = {"SEQN", "CYCLE", "PHQ9_Score", "Depression_Flag", "WTMEC_COMB"}

# Every raw PHQ-9 item shares this prefix and would leak the target
FORBIDDEN_PREFIXES = ("DPQ0",)

# Each saved model, its file name, and the encoding it was trained on
MODEL_FILES = {
    "LogisticRegression": ("logistic_seed42.joblib",      "Nhanes_SVM_Logistic"),
    "RandomForest":       ("random_forest_seed42.joblib", "Nhanes_XGB_Forest"),
    "SVM":                ("svm_seed42.joblib",           "Nhanes_SVM_Logistic"),
    "XGBoost":            ("xgboost_seed42.joblib",       "Nhanes_XGB_Forest"),
    "CatBoost":           ("catboost_seed42.joblib",      "Nhanes_Cat"),
}
```

### Load the Feature and Target Splits

Each model was trained on a different encoding of the same rows. The repository below rebuilds those encodings with the exact column clean-up used in Step 5, and caches each one so a file is read only once.

```python
# Read one raw feature file for an encoding stem and split name
def read_features(stem, split):
    path = os.path.join(SPLIT_DIR, f"{stem}_{split}.csv")
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing split file: {path}")
    return pd.read_csv(path, low_memory=False)


# Return a copy of the frame with the survey weight and redundant features removed
def drop_unwanted_columns(frame):
    to_drop = [c for c in frame.columns if c in DROP_EXACT]
    to_drop += [c for c in frame.columns if any(c.startswith(p) for p in DROP_PREFIX)]
    return frame.drop(columns=sorted(set(to_drop)))


# Stop the notebook if any target-leaking column survived the clean-up
def check_no_leakage(frame, label):
    leaked = [c for c in frame.columns
              if c in FORBIDDEN_COLUMNS or c.startswith(FORBIDDEN_PREFIXES)]
    if leaked:
        raise ValueError(f"{label} still contains leaking columns: {leaked}")
```

```python
# Cache so each CSV is read from disk only once
feature_cache = {}
target_cache = {}


# Return the cleaned feature frame for one encoding stem and split
def load_features(stem, split):
    key = (stem, split)
    if key not in feature_cache:
        cleaned = drop_unwanted_columns(read_features(stem, split))
        check_no_leakage(cleaned, f"{stem}_{split}")
        feature_cache[key] = cleaned
    return feature_cache[key].copy()


# Return the target vector for one split as a flat array
def load_target(split):
    if split not in target_cache:
        path = os.path.join(SPLIT_DIR, f"{TARGET_STEM}_{split}.csv")
        if not os.path.exists(path):
            raise FileNotFoundError(f"Missing target file: {path}")
        target_cache[split] = pd.read_csv(path)["Depression_Flag"].to_numpy()
    return target_cache[split]
```

```python
# Confirm the target splits load and the class balance matches Step 5
# The positive rate stays near 10-11% in all three splits, so the comparison below leads with PR-AUC rather than accuracy.

for split in ["Train", "Val", "Test"]:
    y = load_target(split)
    print(f"{split:<6} n={len(y):>6}  positives={y.sum():>5}  rate={y.mean():.4f}")
```

### Saved Model Artifacts

Every Step 5 notebook saved a small bundle: the fitted estimator, its tuned threshold, and the exact column order it was trained on. The class below wraps one bundle and knows how to prepare a feature frame for it.

```python
# Turn one saved joblib bundle into a plain dictionary this notebook can use
def build_model_entry(name, bundle, stem):
    if "model" not in bundle or "columns" not in bundle:
        raise KeyError(f"{name}: bundle must have 'model' and 'columns' keys")

    return {
        "name": name,
        "model": bundle["model"],
        "columns": list(bundle["columns"]),
        "categorical": list(bundle.get("cat_features", [])),
        "params": bundle.get("params", bundle.get("C", {})),
        "stem": stem,
        "color": MODEL_COLORS.get(name, "#7f8c8d"),

        # Fall back to 0.50 if a Step 5 notebook forgot to save its tuned threshold
        "threshold": float(bundle.get("threshold", 0.50)),
    }


# Load every model file that exists and warn clearly about any that is missing
def load_saved_models():
    entries = []
    missing = []

    for name, (file_name, stem) in MODEL_FILES.items():
        path = os.path.join(MODEL_DIR, file_name)
        if not os.path.exists(path):
            missing.append((name, path))
            continue
        entries.append(build_model_entry(name, joblib.load(path), stem))

    if not entries:
        raise FileNotFoundError(
            f"No saved models found in {MODEL_DIR} - run the Step 5 notebooks first"
        )

    return entries, missing
```

Any model listed as a warning above is simply not in the comparison. The most common cause is a `BASE_DIR` that points outside the project root, which sends the saved file to the wrong folder. Fix the path, re-run that Step 5 notebook, then re-run this notebook and it will be picked up automatically.

```python
# Load whatever Step 5 produced
models, missing_models = load_saved_models()

print(f"Loaded {len(models)} saved model(s):")
for entry in models:
    print(f"  {entry['name']:<20} encoding={entry['stem']:<22} "
          f"threshold={entry['threshold']:.4f}")

for name, path in missing_models:
    print(f"\nWARNING  {name} was not found at {path}")
    print(f"         It is excluded from this comparison.")
    print(f'         If you expected it, check that its Step 5 notebook uses '
          f'BASE_DIR = "." and re-run it.')
```

### Score Every Model

One scoring function is used for every model, so the numbers are directly comparable. Each model is scored twice per split: once at the default 0.50 threshold, and once at the threshold it tuned on validation.

```python
# Line up a feature frame with the column order the model was trained on
def prepare_features(frame, columns, categorical):
    missing = [c for c in columns if c not in frame.columns]
    if missing:
        raise ValueError(
            f"Split is missing {len(missing)} trained columns, first few: {missing[:5]}"
        )

    aligned = frame[columns].copy()

    # CatBoost was trained on string categories, so cast them back before scoring
    for column in categorical:
        aligned[column] = aligned[column].astype(str)

    return aligned


# Return the predicted probability of the positive class for one model and split
def predict_positive_proba(entry, split):
    frame = load_features(entry["stem"], split)
    prepared = prepare_features(frame, entry["columns"], entry["categorical"])
    return entry["model"].predict_proba(prepared)[:, 1]


# Turn one set of predicted probabilities into a full row of metrics
def score_predictions(y_true, proba, threshold):
    pred = (proba >= threshold).astype(int)
    return {
        "PR_AUC":    average_precision_score(y_true, proba),
        "ROC_AUC":   roc_auc_score(y_true, proba),
        "Recall":    recall_score(y_true, pred, zero_division=0),
        "Precision": precision_score(y_true, pred, zero_division=0),
        "F1":        f1_score(y_true, pred, zero_division=0),
        "Accuracy":  accuracy_score(y_true, pred),
        "Threshold": threshold,
    }
```

```python
# Score every model on every split, keeping the probabilities for the curve plots
def score_all_models(entries, splits=("Val", "Test")):
    rows = []
    probabilities = {}

    for entry in entries:
        for split in splits:
            y_true = load_target(split)
            proba = predict_positive_proba(entry, split)

            # Save the probabilities so the charts do not need a second pass
            probabilities[(entry["name"], split)] = (y_true, proba)

            rows.append({
                "Model": entry["name"], "Split": split, "Threshold_Type": "Default 0.50",
                **score_predictions(y_true, proba, 0.50),
            })
            rows.append({
                "Model": entry["name"], "Split": split, "Threshold_Type": "Tuned",
                **score_predictions(y_true, proba, entry["threshold"]),
            })

    column_order = ["Model", "Split", "Threshold_Type", "PR_AUC", "ROC_AUC",
                    "Recall", "Precision", "F1", "Accuracy", "Threshold"]

    return pd.DataFrame(rows)[column_order], probabilities
```

```python
# Score everything that loaded
results, probabilities = score_all_models(models)

print(f"Scored {results['Model'].nunique()} models "
      f"x {results['Split'].nunique()} splits "
      f"x 2 thresholds = {len(results)} rows")
```

### Comparison

#### Test set, at each model's tuned threshold

This is the headline table for the report. Every model is on the same held-out rows, sorted by PR-AUC.


```python
# Validation performance at each model's tuned threshold
val_tuned = (
    results
    .query("Split == 'Val' and Threshold_Type == 'Tuned'")
    .sort_values("PR_AUC", ascending=False)
    .reset_index(drop=True)
)

display(val_tuned.drop(columns="Threshold_Type").round(4))
```

```python
#### Validation to test gap

A large drop from validation to test means the tuning choices did not carry over to unseen data.
```

```python
# Compare PR-AUC on validation against PR-AUC on test for each model
generalization = (
    results
    .query("Threshold_Type == 'Tuned'")
    .pivot(index="Model", columns="Split", values="PR_AUC")
    .rename(columns={"Val": "Val_PR_AUC", "Test": "Test_PR_AUC"})
)

generalization["Gap"] = generalization["Val_PR_AUC"] - generalization["Test_PR_AUC"]
generalization = generalization.sort_values("Test_PR_AUC", ascending=False)

display(generalization.round(4))
```
